Soit $z = e^{j\theta}$, donc $z^{-1} = e^{-j\theta}$.

$$z + z^{-1} = e^{j\theta} + e^{-j\theta} = 2\cos\theta \;\Rightarrow\; \cos\theta = \frac{z+z^{-1}}{2}$$

$$\cos^3\theta = \frac{1}{8}\left(2\cos 3\theta + 6\cos\theta\right)$$

$$\boxed{\cos^3\theta = \frac{1}{4}\cos 3\theta + \frac{3}{4}\cos\theta}$$

C'est une combinaison linéaire de cosinus. On retrouve la formule de triple angle $\cos 3\theta = 4\cos^3\theta - 3\cos\theta$.

### Exercice 4 – Somme de deux sinusoïdes
Soit $x(t) = 3\cos(2\pi t) + 4\sin(2\pi t)$.

1. Mettre $x(t)$ sous la forme $A\cos(2\pi t - \phi)$ avec $A>0$ et $\phi\in[0,2\pi[$.  
2. Vérifier en traçant les deux signaux et leur somme sur l’intervalle $[0,1]$ s.  
3. Généraliser : pour $a\cos(\omega t) + b\sin(\omega t)$, donner les formules de $A$ et $\phi$.

**Réponse** (à compléter) :

```python
# Vérification graphique
import numpy as np
import matplotlib.pyplot as plt

t = np.linspace(0, 1, 1000)
x = 3*np.cos(2*np.pi*t) + 4*np.sin(2*np.pi*t)

# Déterminer A et phi (à calculer)
A = 5.0
phi = np.arctan2(4, 3)  # car cos(phi) = a/A, sin(phi) = b/A
x_phaseur = A * np.cos(2*np.pi*t - phi)

plt.plot(t, x, label='somme')
plt.plot(t, x_phaseur, '--', label='A cos(2πt - φ)')
plt.legend()
plt.grid()
plt.show()
```

**Correction exercice 4**

**1. Forme $A\cos(2\pi t - \phi)$**

On développe : $A\cos(2\pi t - \phi) = A\cos\phi\,\cos(2\pi t) + A\sin\phi\,\sin(2\pi t)$.

Par identification avec $3\cos(2\pi t) + 4\sin(2\pi t)$ :

$$A\cos\phi = 3 \quad\text{et}\quad A\sin\phi = 4$$

Donc :

$$A = \sqrt{3^2 + 4^2} = 5, \qquad \tan\phi = \frac{4}{3}$$

Comme $\cos\phi > 0$ et $\sin\phi > 0$, $\phi$ est dans le premier quadrant :

$$\phi = \arctan\left(\frac{4}{3}\right) \approx 0{,}927 \text{ rad} \approx 53{,}13°$$

$$\boxed{x(t) = 5\cos(2\pi t - 0{,}927)}$$

**3. Généralisation**

Pour $a\cos(\omega t) + b\sin(\omega t)$, la même identification donne $A\cos\phi = a$ et $A\sin\phi = b$ :

$$A = \sqrt{a^2 + b^2}, \qquad \phi = \operatorname{atan2}(b, a)$$

On utilise `atan2` plutôt que $\arctan(b/a)$ pour obtenir le bon quadrant. Si le résultat est négatif, on ajoute $2\pi$ pour avoir $\phi \in [0, 2\pi[$.

```python
import numpy as np
import matplotlib.pyplot as plt

t = np.linspace(0, 1, 1000)

# Les deux composantes et leur somme
x1 = 3*np.cos(2*np.pi*t)
x2 = 4*np.sin(2*np.pi*t)
x = x1 + x2

# Amplitude et phase obtenues en théorie
A = np.sqrt(3**2 + 4**2)
phi = np.arctan2(4, 3)
print(f"A = {A}, phi = {phi:.4f} rad")

x_phaseur = A*np.cos(2*np.pi*t - phi)

plt.plot(t, x1, label='3cos(2πt)')
plt.plot(t, x2, label='4sin(2πt)')
plt.plot(t, x, label='somme')
plt.plot(t, x_phaseur, '--', label='A cos(2πt - φ)')
plt.xlabel('Temps (s)')
plt.legend()
plt.grid()
plt.show()
```

### Exercice 5 – Superposition de trois sinusoïdes
Soit
$$ 
x(t) = \cos(2\pi 100 t) + 2\cos(2\pi 200 t - \pi/4) + 0.5\cos(2\pi 300 t + \pi/3). 
$$  
1. Écrire chaque terme sous forme d’exponentielle complexe.  
2. En déduire l’expression complexe complète de $x(t)$.  
3. Générer et tracer ce signal sur une durée de 20 ms. Que constatez-vous sur la forme d’onde ?

**Réponse** (à compléter) :

```python
# Génération du signal
fs = 10000  # fréquence d'échantillonnage
t = np.linspace(0, 0.02, int(0.02*fs))

x = np.cos(2*np.pi*100*t) + 2*np.cos(2*np.pi*200*t - np.pi/4) + 0.5*np.cos(2*np.pi*300*t + np.pi/3)

plt.plot(t, x)
plt.xlabel('Temps (s)')
plt.ylabel('Amplitude')
plt.title('Superposition de trois sinusoïdes')
plt.grid()
plt.show()
```

**Correction exercice 5**

**1. Chaque terme sous forme d'exponentielle complexe**

On utilise $\cos(\alpha) = \frac{e^{j\alpha} + e^{-j\alpha}}{2}$ :

$$\cos(2\pi 100 t) = \frac{1}{2}e^{j2\pi 100 t} + \frac{1}{2}e^{-j2\pi 100 t}$$

$$2\cos\left(2\pi 200 t - \frac{\pi}{4}\right) = e^{-j\pi/4}e^{j2\pi 200 t} + e^{j\pi/4}e^{-j2\pi 200 t}$$

$$0{,}5\cos\left(2\pi 300 t + \frac{\pi}{3}\right) = \frac{1}{4}e^{j\pi/3}e^{j2\pi 300 t} + \frac{1}{4}e^{-j\pi/3}e^{-j2\pi 300 t}$$

**2. Expression complexe complète**

$$x(t) = \frac{1}{2}e^{j2\pi 100 t} + e^{-j\pi/4}e^{j2\pi 200 t} + \frac{1}{4}e^{j\pi/3}e^{j2\pi 300 t} + \text{(conjugués : fréquences } -100, -200, -300 \text{ Hz)}$$

Ou, de façon équivalente, avec les phaseurs complexes $X_1 = 1$, $X_2 = 2e^{-j\pi/4}$, $X_3 = 0{,}5e^{j\pi/3}$ :

$$x(t) = \operatorname{Re}\left\{X_1 e^{j2\pi 100 t} + X_2 e^{j2\pi 200 t} + X_3 e^{j2\pi 300 t}\right\}$$

**3. Observation**

Les fréquences 100, 200 et 300 Hz sont des multiples entiers de $f_0 = 100$ Hz. Le signal est donc **périodique** de période $T_0 = 1/f_0 = 10$ ms. Sur 20 ms, on observe deux périodes identiques.

La forme d'onde n'est pas une sinusoïde. C'est une forme plus complexe, qui vient de la superposition de l'harmonique 1 (fondamental), de l'harmonique 2 et de l'harmonique 3. Chaque harmonique a sa propre amplitude et sa propre phase, ce qui détermine la forme du motif qui se répète.

```python
import numpy as np
import matplotlib.pyplot as plt

# Génération du signal
fs = 10000  # fréquence d'échantillonnage (Hz)
t = np.arange(0, 0.02, 1/fs)   # pas exact de 1/fs, 200 échantillons

x = (np.cos(2*np.pi*100*t)
     + 2*np.cos(2*np.pi*200*t - np.pi/4)
     + 0.5*np.cos(2*np.pi*300*t + np.pi/3))

# Même signal calculé avec les exponentielles complexes (vérification)
x_cplx = np.real(1*np.exp(1j*2*np.pi*100*t)
                 + 2*np.exp(-1j*np.pi/4)*np.exp(1j*2*np.pi*200*t)
                 + 0.5*np.exp(1j*np.pi/3)*np.exp(1j*2*np.pi*300*t))

print("Écart max :", np.max(np.abs(x - x_cplx)))

plt.plot(t, x, label='cosinus')
plt.plot(t, x_cplx, '--', label='exponentielles')
plt.xlabel('Temps (s)')
plt.ylabel('Amplitude')
plt.title('Superposition de trois sinusoïdes')
plt.legend()
plt.grid()
plt.show()
```

### Exercice 6 – Battements
Deux signaux de fréquences proches :  
$
x_1(t) = \cos(2\pi f_1 t)
$ et 
$
x_2(t) = \cos(2\pi f_2 t)
$.  
Montrer que leur somme peut s’écrire  
$$ 
2\cos(2\pi f_m t)\cos(2\pi f_d t) 
$$  
avec $f_m = \frac{f_1+f_2}{2}$ et $f_d = \frac{f_1-f_2}{2}$.  
Pour $f_1 = 10$ Hz et $f_2 = 12$ Hz, tracer la somme et observer l’enveloppe. Expliquer le phénomène de battement.

**Démonstration** (à compléter) :

```python
f1 = 10
f2 = 12
t = np.linspace(0, 1, 2000)

x1 = np.cos(2*np.pi*f1*t)
x2 = np.cos(2*np.pi*f2*t)
x_sum = x1 + x2

fm = (f1+f2)/2
fd = (f1-f2)/2
x_env = 2*np.cos(2*np.pi*fd*t)  # enveloppe
x_product = x_env * np.cos(2*np.pi*fm*t)

plt.plot(t, x_sum, label='Somme')
plt.plot(t, x_env, '--', label='Enveloppe')
plt.plot(t, -x_env, '--', label='-Enveloppe')
plt.xlabel('Temps (s)')
plt.legend()
plt.grid()
plt.show()

# Explication : la somme de deux sinusoïdes de fréquences proches donne une oscillation rapide (fm) modulée par une enveloppe lente (fd).
```

**Coorection exercice 6**

On utilise la formule de somme en produit : $\cos a + \cos b = 2\cos\left(\frac{a+b}{2}\right)\cos\left(\frac{a-b}{2}\right)$.

Avec $a = 2\pi f_1 t$ et $b = 2\pi f_2 t$ :

$$x_1(t) + x_2(t) = 2\cos\left(2\pi\frac{f_1+f_2}{2}\,t\right)\cos\left(2\pi\frac{f_1-f_2}{2}\,t\right) = 2\cos(2\pi f_m t)\cos(2\pi f_d t)$$

avec $f_m = \frac{f_1+f_2}{2}$ et $f_d = \frac{f_1-f_2}{2}$.

On peut aussi le montrer avec Euler : on écrit $f_1 = f_m + f_d$ et $f_2 = f_m - f_d$, donc

$$\cos(2\pi(f_m+f_d)t) + \cos(2\pi(f_m-f_d)t) = 2\cos(2\pi f_m t)\cos(2\pi f_d t).$$

**Application numérique** : $f_1 = 10$ Hz et $f_2 = 12$ Hz

$$f_m = 11 \text{ Hz}, \qquad f_d = -1 \text{ Hz}$$

Comme le cosinus est pair, $\cos(2\pi f_d t) = \cos(2\pi t)$ (fréquence de 1 Hz).

**Explication du phénomène de battement**

La somme est une oscillation **rapide** à la fréquence moyenne $f_m = 11$ Hz, dont l'amplitude est modulée par une **enveloppe lente** $2\cos(2\pi f_d t)$ de fréquence $|f_d| = 1$ Hz.

Les deux signaux sont tantôt en phase (ils s'additionnent, l'amplitude atteint 2), tantôt en opposition de phase (ils s'annulent, l'amplitude est nulle). L'amplitude passe par un maximum à chaque fois que l'enveloppe atteint $\pm 2$, soit deux fois par période de l'enveloppe.

La **fréquence de battement**, c'est-à-dire le nombre de maxima d'amplitude par seconde, vaut donc :

$$f_{batt} = |f_1 - f_2| = 2 \text{ Hz}$$

On entend ce phénomène quand deux diapasons de fréquences proches sonnent ensemble : le son semble « pulser » 2 fois par seconde.

```python
import numpy as np
import matplotlib.pyplot as plt

f1 = 10
f2 = 12
t = np.linspace(0, 1, 2000)

x1 = np.cos(2*np.pi*f1*t)
x2 = np.cos(2*np.pi*f2*t)
x_sum = x1 + x2

fm = (f1 + f2)/2
fd = (f1 - f2)/2
x_env = 2*np.cos(2*np.pi*fd*t)           # enveloppe lente
x_product = x_env*np.cos(2*np.pi*fm*t)   # forme "produit" de la démonstration

# Vérification : somme et produit doivent être identiques
print("Écart max :", np.max(np.abs(x_sum - x_product)))

plt.figure(figsize=(12, 4))
plt.plot(t, x_sum, label='Somme')
plt.plot(t, x_env, '--', label='Enveloppe')
plt.plot(t, -x_env, '--', label='-Enveloppe')
plt.xlabel('Temps (s)')
plt.legend()
plt.grid()
plt.show()
```

### Exercice 7 – Signal exponentiel complexe et parties réelle/imaginaire
Générer le signal complexe $z(t) = e^{j(2\pi 5 t + \pi/3)}$.  
Tracer simultanément la partie réelle, la partie imaginaire, et le module (qui doit être constant). Ajouter le tracé de la phase (argument) au cours du temps. Que remarquez-vous sur la phase ?

**Code** :

```python
import numpy as np
import matplotlib.pyplot as plt

t = np.linspace(0, 1, 1000)
z = np.exp(1j * (2*np.pi*5*t + np.pi/3))

plt.figure(figsize=(14, 6))
plt.subplot(2, 1, 1)
plt.plot(t, np.real(z), label='Partie réelle')
plt.plot(t, np.imag(z), label='Partie imaginaire')
plt.plot(t, np.abs(z), label='Module')
plt.legend()
plt.grid()

plt.subplot(2, 1, 2)
plt.plot(t, np.unwrap(np.angle(z)), label='Phase')
plt.xlabel('Temps (s)')
plt.ylabel('Phase (rad)')
plt.grid()
plt.legend()
plt.show()

# La phase augmente linéairement avec le temps (pente = 2π*5).
```

**Correction exercice 7**

Le signal est $z(t) = e^{j(2\pi 5 t + \pi/3)}$. D'après Euler :

$$z(t) = \cos\left(2\pi 5 t + \frac{\pi}{3}\right) + j\,\sin\left(2\pi 5 t + \frac{\pi}{3}\right)$$

- **Partie réelle** : $\operatorname{Re}\{z(t)\} = \cos(2\pi 5 t + \pi/3)$
- **Partie imaginaire** : $\operatorname{Im}\{z(t)\} = \sin(2\pi 5 t + \pi/3)$
- **Module** : $|z(t)| = \sqrt{\cos^2 + \sin^2} = 1$, donc constant.
- **Phase** : $\arg z(t) = 2\pi 5 t + \pi/3$

**Observations**

- La partie réelle et la partie imaginaire sont deux sinusoïdes de même fréquence (5 Hz), décalées d'un quart de période ($\pi/2$).
- Le module reste constant égal à 1 : le point $z(t)$ décrit un cercle de rayon 1 dans le plan complexe.
- La phase **augmente linéairement** avec le temps. Sa pente vaut $2\pi \times 5 = 10\pi$ rad/s, c'est-à-dire $2\pi f$ (la pulsation $\omega$). Elle part de $\pi/3$ à $t = 0$.
- Sans `np.unwrap`, `np.angle` renvoie une phase repliée dans $]-\pi, \pi]$ et on obtient une courbe en dents de scie. Avec `np.unwrap`, on retrouve la droite continue.

```python
import numpy as np
import matplotlib.pyplot as plt

t = np.linspace(0, 1, 1000)
z = np.exp(1j * (2*np.pi*5*t + np.pi/3))

plt.figure(figsize=(14, 6))

plt.subplot(2, 1, 1)
plt.plot(t, np.real(z), label='Partie réelle')
plt.plot(t, np.imag(z), label='Partie imaginaire')
plt.plot(t, np.abs(z), label='Module')
plt.legend()
plt.grid()

phase = np.unwrap(np.angle(z))
plt.subplot(2, 1, 2)
plt.plot(t, phase, label='Phase')
plt.xlabel('Temps (s)')
plt.ylabel('Phase (rad)')
plt.grid()
plt.legend()
plt.tight_layout()
plt.show()

# Vérification : phase initiale et pente
print("Phase initiale :", phase[0], "(pi/3 =", np.pi/3, ")")
print("Pente :", (phase[-1] - phase[0]) / (t[-1] - t[0]), "(10*pi =", 10*np.pi, ")")
```

### Exercice 8 – Modulation d’amplitude
Un signal modulé en amplitude s’écrit  
$$ 
s(t) = A_c \big[1 + m \cos(2\pi f_m t)\big] \cos(2\pi f_c t).
$$  
Générer ce signal pour $f_c = 100$ Hz, $f_m = 5$ Hz, $m = 0.5$, $A_c = 1$.  
Tracer le signal sur 0,2 s, puis calculer sa transformée de Fourier (FFT) pour visualiser les raies spectrales (porteuse et bandes latérales).  
Expliquer pourquoi il y a trois composantes fréquentielles.

**Code** :

```python
fs = 1000  # Hz
t = np.linspace(0, 0.2, int(0.2*fs))
fc = 100
fm = 5
m = 0.5
Ac = 1

s = Ac * (1 + m*np.cos(2*np.pi*fm*t)) * np.cos(2*np.pi*fc*t)

plt.figure(figsize=(14, 5))
plt.subplot(1, 2, 1)
plt.plot(t, s)
plt.xlabel('Temps (s)')
plt.title('Signal modulé en amplitude')
plt.grid()

# FFT
N = len(s)
freqs = np.fft.fftfreq(N, 1/fs)
spectre = np.fft.fft(s)

plt.subplot(1, 2, 2)
plt.stem(freqs[:N//2], np.abs(spectre[:N//2]) / N)
plt.xlim(0, 150)
plt.xlabel('Fréquence (Hz)')
plt.ylabel('Amplitude')
plt.title('Spectre (3 raies)')
plt.grid()
plt.show()

# Explication : le produit de cosinus donne des composantes à fc, fc+fm, fc-fm.
```

**Correction exercice 8**

Le signal modulé est :

$$s(t) = A_c\big[1 + m\cos(2\pi f_m t)\big]\cos(2\pi f_c t)$$

On développe, puis on utilise $\cos a\cos b = \frac{1}{2}\left[\cos(a+b) + \cos(a-b)\right]$ :

$$s(t) = A_c\cos(2\pi f_c t) + \frac{A_c m}{2}\cos\big(2\pi (f_c+f_m) t\big) + \frac{A_c m}{2}\cos\big(2\pi (f_c-f_m) t\big)$$

**Pourquoi trois composantes ?**

Le produit de la porteuse (à $f_c$) par le message (à $f_m$) crée de nouvelles fréquences : la somme et la différence. On obtient donc trois raies :

| Composante | Fréquence | Amplitude |
|---|---|---|
| Bande latérale inférieure | $f_c - f_m = 95$ Hz | $\frac{A_c m}{2} = 0{,}25$ |
| Porteuse | $f_c = 100$ Hz | $A_c = 1$ |
| Bande latérale supérieure | $f_c + f_m = 105$ Hz | $\frac{A_c m}{2} = 0{,}25$ |

La porteuse ne porte pas d'information. L'information du message se trouve dans les deux bandes latérales, symétriques autour de $f_c$.

**Remarque sur la FFT**

Chaque cosinus donne deux raies (une en $+f$ et une en $-f$), chacune d'amplitude moitié. En ne gardant que les fréquences positives et en divisant par $N$, on lit donc $0{,}5$ pour la porteuse et $0{,}125$ pour chaque bande latérale (la moitié des amplitudes du tableau).

```python
import numpy as np
import matplotlib.pyplot as plt

fs = 1000  # fréquence d'échantillonnage (Hz)
t = np.arange(0, 0.2, 1/fs)   # 200 échantillons, pas exact 1/fs
fc = 100
fm = 5
m = 0.5
Ac = 1

s = Ac * (1 + m*np.cos(2*np.pi*fm*t)) * np.cos(2*np.pi*fc*t)

plt.figure(figsize=(14, 5))

plt.subplot(1, 2, 1)
plt.plot(t, s)
plt.xlabel('Temps (s)')
plt.title('Signal modulé en amplitude')
plt.grid()

# FFT
N = len(s)
freqs = np.fft.fftfreq(N, 1/fs)
spectre = np.fft.fft(s)

plt.subplot(1, 2, 2)
plt.stem(freqs[:N//2], np.abs(spectre[:N//2]) / N)
plt.xlim(0, 150)
plt.xlabel('Fréquence (Hz)')
plt.ylabel('Amplitude')
plt.title('Spectre (3 raies)')
plt.grid()
plt.tight_layout()
plt.show()
```

### Exercice 9 – Repliement de spectre (aliasing) – approche numérique
Échantillonner le signal $x(t) = \cos(2\pi 30 t)$ avec une fréquence d’échantillonnage $f_e = 20$ Hz (donc sous-échantillonnage).  
Générer les échantillons pour $t = 0, 0.05, 0.10, \dots$ et les tracer.  
Reconstruire le signal à partir des échantillons en traçant la droite reliant les points (ou en utilisant une interpolation).  
Quelle fréquence apparente observez-vous ? Justifier à l’aide de la formule d’aliasing.

**Code** :

```python
fe = 20
f = 30
t_cont = np.linspace(0, 0.5, 1000)
x_cont = np.cos(2*np.pi*f*t_cont)

# Échantillonnage
te = 1/fe
t_ech = np.arange(0, 0.5, te)
x_ech = np.cos(2*np.pi*f*t_ech)

plt.plot(t_cont, x_cont, label='Signal original (30 Hz)')
plt.stem(t_ech, x_ech, linefmt='r-', markerfmt='ro', basefmt=' ', label='Échantillons (20 Hz)')

# Reconstruction linéaire (simple interpolation)
t_recon = np.linspace(0, 0.5, 1000)
x_recon = np.interp(t_recon, t_ech, x_ech)
plt.plot(t_recon, x_recon, '--', label='Reconstruction')

plt.legend()
plt.grid()
plt.show()

# La fréquence apparente est |fe - f| = 10 Hz. En effet, l'échantillonnage à 20 Hz donne un alias à 10 Hz.
```

**Correction exercice 9**

Le signal est $x(t) = \cos(2\pi \cdot 30\, t)$, échantillonné à $f_e = 20$ Hz, soit $T_e = 0{,}05$ s.

Les échantillons valent :

$$x[n] = \cos(2\pi \cdot 30 \cdot n T_e) = \cos\left(2\pi \frac{30}{20} n\right) = \cos(3\pi n) = (-1)^n$$

Les échantillons alternent donc $+1, -1, +1, -1, \dots$

**Condition de Shannon**

Pour ne pas avoir de repliement, il faut $f_e > 2 f_{max}$, soit ici $f_e > 60$ Hz. Avec $f_e = 20$ Hz, la condition n'est pas respectée : il y a **sous-échantillonnage**.

**Fréquence apparente (formule d'aliasing)**

Deux fréquences $f$ et $f + k f_e$ ($k \in \mathbb{Z}$) donnent exactement les mêmes échantillons, car $e^{j2\pi k f_e n T_e} = e^{j2\pi k n} = 1$. La fréquence apparente est donc :

$$f_{app} = |f - k f_e| \quad\text{avec } k \text{ choisi pour que } f_{app} \in \left[0, \tfrac{f_e}{2}\right]$$

Ici, avec $k = 2$ : $f_{app} = |30 - 2 \times 20| = 10$ Hz (de même, avec $k = 1$, on trouve $|30 - 20| = 10$ Hz).

Vérification : $\cos(2\pi \cdot 10 \cdot n T_e) = \cos(\pi n) = (-1)^n$. On retrouve exactement les mêmes échantillons que pour le signal à 30 Hz.

**Observation**

Le signal reconstruit à partir des échantillons oscille à **10 Hz**, et non à 30 Hz. Comme 10 Hz $= f_e/2$, il y a seulement 2 échantillons par période, et l'interpolation linéaire donne une forme en **triangle** plutôt qu'une sinusoïde. L'information sur la fréquence d'origine est perdue : on ne peut plus distinguer 30 Hz de 10 Hz.

```python
import numpy as np
import matplotlib.pyplot as plt

fe = 20
f = 30
t_cont = np.linspace(0, 0.5, 1000)
x_cont = np.cos(2*np.pi*f*t_cont)

# Échantillonnage
te = 1/fe
t_ech = np.arange(0, 0.5, te)
x_ech = np.cos(2*np.pi*f*t_ech)

# Alias à 10 Hz : doit passer par les mêmes échantillons
f_app = abs(f - 2*fe)
x_alias = np.cos(2*np.pi*f_app*t_cont)
print("Fréquence apparente :", f_app, "Hz")
print("Écart max sur les échantillons :", np.max(np.abs(x_ech - np.cos(2*np.pi*f_app*t_ech))))

plt.figure(figsize=(12, 5))
plt.plot(t_cont, x_cont, label='Signal original (30 Hz)')
plt.plot(t_cont, x_alias, ':', label=f'Alias ({f_app} Hz)')
plt.stem(t_ech, x_ech, linefmt='r-', markerfmt='ro', basefmt=' ', label='Échantillons (20 Hz)')

# Reconstruction linéaire (simple interpolation)
x_recon = np.interp(t_cont, t_ech, x_ech)
plt.plot(t_cont, x_recon, '--', label='Reconstruction')

plt.xlabel('Temps (s)')
plt.legend()
plt.grid()
plt.show()
```

### Exercice 10 – Synthèse d’un signal carré par séries de Fourier
Un signal carré de période $T$ et d’amplitude 1 peut être décomposé en série de Fourier :  
$$
x(t) = \frac{4}{\pi} \sum_{k=0}^{N} \frac{1}{2k+1} \sin\bigl(2\pi (2k+1) f_0 t\bigr).
$$  
Écrire une fonction Python qui génère ce signal pour un nombre d’harmoniques $N$ donné.  
Tester pour $N=1, 5, 20, 50$ et tracer les signaux correspondants.  
Observer le phénomène de Gibbs (oscillations près des discontinuités).  
Comparer avec la fonction `square` de `scipy.signal`.

**Code** :

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import signal

def carre_fourier(t, f0, N):
    """Génère un signal carré par synthèse de Fourier avec N harmoniques impaires."""
    x = np.zeros_like(t)
    for k in range(N):
        n = 2*k + 1
        x += (1/n) * np.sin(2*np.pi*n*f0*t)
    return (4/np.pi) * x

f0 = 1  # Hz
t = np.linspace(0, 2, 1000)

fig, axes = plt.subplots(2, 2, figsize=(12, 8))
Ns = [1, 5, 20, 50]

for ax, N in zip(axes.flat, Ns):
    x = carre_fourier(t, f0, N)
    ax.plot(t, x, label=f'N={N}')
    ax.grid()
    ax.legend()

plt.suptitle('Synthèse d\'un signal carré par séries de Fourier')
plt.tight_layout()
plt.show()

# Comparaison avec square
x_square = signal.square(2*np.pi*f0*t)
plt.plot(t, x_square, label='square (scipy)')
plt.plot(t, carre_fourier(t, f0, 50), '--', label='N=50')
plt.legend()
plt.grid()
plt.show()
```

**Correction exercice 10**

Un signal carré de période $T = 1/f_0$ et d'amplitude 1 se décompose en série de Fourier :

$$x(t) = \frac{4}{\pi}\sum_{k=0}^{N}\frac{1}{2k+1}\sin\big(2\pi(2k+1)f_0 t\big)$$

- Il ne contient que des **harmoniques impaires** ($f_0, 3f_0, 5f_0, \dots$), car le signal carré est impair et à symétrie demi-période.
- L'amplitude de l'harmonique $n = 2k+1$ décroît en $\frac{4}{\pi n}$ : le fondamental vaut $\frac{4}{\pi} \approx 1{,}27$, la 3e harmonique $\frac{4}{3\pi} \approx 0{,}42$, etc.
- Le sinus est impair, ce qui est cohérent avec un carré qui vaut $+1$ sur la première demi-période et $-1$ sur la seconde.

**Observations**

- Plus $N$ augmente, plus la somme se rapproche du signal carré : les plateaux deviennent plats et les fronts plus raides.
- **Phénomène de Gibbs** : près des discontinuités, il reste un dépassement (« overshoot ») d'environ **9 % du saut**. Le saut vaut 2 (de $-1$ à $+1$), donc le pic monte à environ $1{,}18$.
- Quand $N$ augmente, ce dépassement **ne diminue pas** : il se **rétrécit** et se rapproche de la discontinuité, mais sa hauteur reste la même. C'est la conséquence de la convergence non uniforme de la série de Fourier en un point de discontinuité.
- La comparaison avec `scipy.signal.square` montre que $N = 50$ donne une très bonne approximation, sauf aux fronts (Gibbs).

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import signal

def carre_fourier(t, f0, N):
    """Génère un signal carré par synthèse de Fourier avec N harmoniques impaires."""
    x = np.zeros_like(t)
    for k in range(N):          # k = 0 .. N-1 -> N harmoniques : 1, 3, 5, ...
        n = 2*k + 1
        x += (1/n) * np.sin(2*np.pi*n*f0*t)
    return (4/np.pi) * x

f0 = 1  # Hz
t = np.linspace(0, 2, 2000)
x_square = signal.square(2*np.pi*f0*t)

fig, axes = plt.subplots(2, 2, figsize=(12, 8))
Ns = [1, 5, 20, 50]

for ax, N in zip(axes.flat, Ns):
    x = carre_fourier(t, f0, N)
    ax.plot(t, x, label=f'N={N}')
    ax.grid()
    ax.legend()
    # Dépassement maximal (Gibbs) et erreur quadratique moyenne
    rms = np.sqrt(np.mean((x - x_square)**2))
    print(f"N={N:3d} | max = {x.max():.3f} | erreur RMS = {rms:.3f}")

plt.suptitle("Synthèse d'un signal carré par séries de Fourier")
plt.tight_layout()
plt.show()

# Comparaison avec square
plt.figure(figsize=(10, 4))
plt.plot(t, x_square, label='square (scipy)')
plt.plot(t, carre_fourier(t, f0, 50), '--', label='N=50')
plt.xlabel('Temps (s)')
plt.legend()
plt.grid()
plt.show()
```

### Exercice 11 – Phase et décalage temporel
Montrer que $A\cos(2\pi f t - \phi)$ est équivalent à un décalage temporel. Quel est le décalage en secondes ?  
Pour $A=1, f=2$ Hz, $\phi = \pi/2$, calculer le décalage. Tracer le signal non décalé et le signal décalé pour vérifier.

**Démonstration** : $\cos(2\pi f (t - \Delta t)) = \cos(2\pi f t - 2\pi f \Delta t)$ donc $\phi = 2\pi f \Delta t$ et $\Delta t = \phi/(2\pi f)$.

**Code** :

```python
f = 2
phi = np.pi/2
A = 1
delta_t = phi / (2*np.pi*f)
print(f"Décalage temporel : {delta_t:.4f} s")

t = np.linspace(0, 1, 1000)
x_original = A * np.cos(2*np.pi*f*t)
x_decale = A * np.cos(2*np.pi*f*(t - delta_t))
x_phase = A * np.cos(2*np.pi*f*t - phi)

plt.plot(t, x_original, label='Original (phi=0)')
plt.plot(t, x_decale, '--', label='Décalé temporel')
plt.plot(t, x_phase, ':', label='Phase -π/2')
plt.legend()
plt.grid()
plt.show()
```

**Correction exercice 11**

On factorise la pulsation dans l'argument du cosinus :

$$A\cos(2\pi f t - \phi) = A\cos\left(2\pi f\left(t - \frac{\phi}{2\pi f}\right)\right)$$

En posant $\Delta t = \dfrac{\phi}{2\pi f}$, on obtient :

$$A\cos(2\pi f t - \phi) = A\cos\big(2\pi f (t - \Delta t)\big)$$

C'est le signal $A\cos(2\pi f t)$ **retardé** de $\Delta t$ secondes (si $\phi > 0$). Un déphasage $\phi$ est donc équivalent à un décalage temporel :

$$\boxed{\Delta t = \frac{\phi}{2\pi f}}$$

**Application numérique** : $A = 1$, $f = 2$ Hz, $\phi = \pi/2$

$$\Delta t = \frac{\pi/2}{2\pi \times 2} = \frac{1}{8} = 0{,}125 \text{ s}$$

Vérification : la période vaut $T = 1/f = 0{,}5$ s et $\phi = \pi/2$ correspond à un quart de période, soit $T/4 = 0{,}125$ s. ✓

**Observation**

Le signal décalé est identique au signal avec la phase $-\pi/2$ : les deux courbes se superposent. Elles sont en retard de 0,125 s par rapport au signal d'origine. Ici, $\cos(2\pi f(t - \Delta t)) = \sin(2\pi f t)$.

```python
import numpy as np
import matplotlib.pyplot as plt

f = 2
phi = np.pi/2
A = 1
delta_t = phi / (2*np.pi*f)
print(f"Décalage temporel : {delta_t:.4f} s")

t = np.linspace(0, 1, 1000)
x_original = A * np.cos(2*np.pi*f*t)
x_decale = A * np.cos(2*np.pi*f*(t - delta_t))
x_phase = A * np.cos(2*np.pi*f*t - phi)

# Vérification : le signal décalé et le signal déphasé sont identiques
print("Écart max :", np.max(np.abs(x_decale - x_phase)))

plt.figure(figsize=(10, 4))
plt.plot(t, x_original, label='Original (phi=0)')
plt.plot(t, x_decale, '--', label='Décalé temporel')
plt.plot(t, x_phase, ':', linewidth=3, label='Phase -π/2')
plt.xlabel('Temps (s)')
plt.legend()
plt.grid()
plt.show()
```

### Exercice 12 – Échantillonnage d’une exponentielle complexe
Soit $z[n] = e^{j\omega_0 n}$. Montrer que si $\omega_0$ est un multiple rationnel de $2\pi$, le signal est périodique.  
Pour $\omega_0 = 2\pi/3$, déterminer la période fondamentale.  
Pour $\omega_0 = 1$ rad/échantillon, le signal est-il périodique ? Justifier.  
Générer les deux signaux sur 50 échantillons et observer leur comportement.

**Démonstration** : La périodicité exige $\omega_0 N = 2\pi k$ avec $k$ entier, donc $\omega_0/(2\pi) = k/N$ est rationnel.

**Code** :

```python
n = np.arange(0, 50)

omega1 = 2*np.pi/3
z1 = np.exp(1j * omega1 * n)

omega2 = 1
z2 = np.exp(1j * omega2 * n)

plt.figure(figsize=(14, 5))

plt.subplot(1, 2, 1)
plt.stem(n, np.real(z1), basefmt=' ')
plt.title(f'$\\omega_0 = 2\pi/3$ (périodique, période 3)')

plt.subplot(1, 2, 2)
plt.stem(n, np.real(z2), basefmt=' ')
plt.title('$\\omega_0 = 1$ (non périodique)')

plt.tight_layout()
plt.show()

# Vérification : pour omega1, la période est 3 car 2pi/3 * 3 = 2pi.
```

**Correction exercice 12**

Un signal discret $z[n]$ est périodique de période $N$ (entier $\geq 1$) si $z[n+N] = z[n]$ pour tout $n$.

Pour $z[n] = e^{j\omega_0 n}$ :

$$z[n+N] = e^{j\omega_0 n}\,e^{j\omega_0 N} = z[n] \iff e^{j\omega_0 N} = 1 \iff \omega_0 N = 2\pi k, \quad k \in \mathbb{Z}$$

Donc $\dfrac{\omega_0}{2\pi} = \dfrac{k}{N}$ doit être un **nombre rationnel**.

- Si $\dfrac{\omega_0}{2\pi} = \dfrac{k}{N}$ avec $k$ et $N$ premiers entre eux, le signal est périodique et sa **période fondamentale** est $N$.
- Si $\dfrac{\omega_0}{2\pi}$ est irrationnel, aucun entier $N$ ne convient : le signal n'est **pas périodique**.

**Cas $\omega_0 = 2\pi/3$**

$$\frac{\omega_0}{2\pi} = \frac{1}{3} \;\Rightarrow\; k = 1,\; N = 3$$

Le signal est périodique, de période fondamentale $N = 3$ échantillons. Vérification : $\omega_0 \times 3 = 2\pi$, donc $e^{j\omega_0 \cdot 3} = 1$.

Les valeurs de $\operatorname{Re}\{z[n]\} = \cos\left(\frac{2\pi n}{3}\right)$ se répètent : $1,\; -\tfrac{1}{2},\; -\tfrac{1}{2},\; 1,\; -\tfrac{1}{2},\; -\tfrac{1}{2},\; \dots$

**Cas $\omega_0 = 1$ rad/échantillon**

$$\frac{\omega_0}{2\pi} = \frac{1}{2\pi}$$

Ce nombre est **irrationnel** (car $\pi$ est irrationnel), donc le signal **n'est pas périodique** : aucun entier $N$ ne vérifie $N = 2\pi k$.

**Observation**

Bien que $z_2[n]$ soit issu d'une sinusoïde continue périodique (de période $2\pi$ s), son échantillonnage ne se répète jamais exactement : le motif des échantillons est différent à chaque oscillation. Une sinusoïde continue est toujours périodique, mais **sa version discrète ne l'est que si $\omega_0/2\pi$ est rationnel**.

```python
import numpy as np
import matplotlib.pyplot as plt

n = np.arange(0, 50)

omega1 = 2*np.pi/3
z1 = np.exp(1j * omega1 * n)

omega2 = 1
z2 = np.exp(1j * omega2 * n)

def periode(z, tol=1e-6):
    """Plus petit N >= 1 tel que z[n+N] = z[n] (None si aucun N < len(z)/2)."""
    for N in range(1, len(z)//2):
        if np.allclose(z[N:], z[:-N], atol=tol):
            return N
    return None

print("Période de z1 :", periode(z1))   # attendu : 3
print("Période de z2 :", periode(z2))   # attendu : None

plt.figure(figsize=(14, 5))

plt.subplot(1, 2, 1)
plt.stem(n, np.real(z1), basefmt=' ')
plt.title(r'$\omega_0 = 2\pi/3$ (périodique, période 3)')
plt.xlabel('n')
plt.grid()

plt.subplot(1, 2, 2)
plt.stem(n, np.real(z2), basefmt=' ')
plt.title(r'$\omega_0 = 1$ (non périodique)')
plt.xlabel('n')
plt.grid()

plt.tight_layout()
plt.show()
```
